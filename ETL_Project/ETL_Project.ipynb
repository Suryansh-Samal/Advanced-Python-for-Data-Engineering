```python
import json
import pandas as pd #importing required modules
```

```python
f = open(r'C:\Users\Suryansh\Documents\gitsql\Python-for-DE-Advanced\10_Pandas\Dataset\orders_ETL.json')
data = json.load(f)    #cant convert the json document directly in the dataframe becuz of messy data
data
```

```python
order_id = []  #creating lists to store ideal data in each list
order_date = []  
total_amount = []
quantity = []
customer_id = []
product_id = []
product_name = []
category = []
price = []
cust_id = []
customer_name = []
email = []
address = []

for orders in data:   #using loop to append data
    cust_id.append(orders['customer']['customer_id'])  #data for customer table
    customer_name.append(orders['customer']['name'])
    email.append(orders['customer']['email'])          
    address.append(orders['customer']['address'])
    for product in orders['products']: #loop for orders and products table
        product_id.append(product['product_id']) 
        quantity.append(product['quantity'])
        order_id.append(orders['order_id'])
        order_date.append(orders['order_date'])
        total_amount.append(orders['total_amount'])
        customer_id.append(orders['customer']['customer_id'])
        product_name.append(product['name'])
        category.append(product['category'])
        price.append(product['price'])

df_orders = pd.DataFrame({           #creating dataframe for orders table
    'customer_id': customer_id,    
    'order_id' :order_id,
    'product_id': product_id,
    'order_date': order_date,
    'total_amount': total_amount,
    'quantity': quantity
})

df_products = pd.DataFrame({          #creating dataframe for products table
    'product_id' : product_id,
    'product_name' : product_name,
    'category': category,
    'price': price
})

df_customers = pd.DataFrame({        #creating dataframe for customers table
    'customer_id': cust_id,
    'customer_name' :customer_name,
    'email': email,
    'address': address
})

df_customers

```

```python
df_customers.drop_duplicates(subset='customer_id', keep='last', inplace=True)  #removing duplicates from products and customers table
df_customers.reset_index(drop=True, inplace=True)
df_products.drop_duplicates(subset='product_id', keep='last', inplace=True)
df_products.reset_index(drop=True, inplace=True)
df_products
```
