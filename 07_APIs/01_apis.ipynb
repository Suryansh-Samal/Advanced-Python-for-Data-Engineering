```python
import requests 
import json
```

# GET

```python
url = 'https://jsonplaceholder.typicode.com/posts/3'

response = requests.get(url)

#Status code
print(response.status_code)

#URL
print(response.url)

#Headers
print(response.headers)

#Response body
print(response.text)
print(response.json())
data = response.json()

#Parse JSON response
print(json.dumps(data, indent=2))
```

# POST

```python
url = 'https://jsonplaceholder.typicode.com/posts'

data = {
    'userID': 1,
    'title': 'My New Post',
    'body': 'This is the content of my new post.'
}

#Method 1 using json parameter
response = requests.post(url, json=data)
print(response.status_code) #Code 201 signifies that the resource has been created successfully
print(json.dumps(response.json(), indent=2))

#Method 2 using data parameter
data_form = {
    'userID': 1,
    'title': 'My 2nd Post',
    'body': 'This is the content of my 2nd post.'
}

response = requests.post(url, data=data_form)
print(response.status_code)
print(json.dumps(response.json(), indent=2))
```

# Query Parameters

```python
url = 'https://jsonplaceholder.typicode.com/posts'

params = {
    "userId": 1,
    "id" : 1
    }

response = requests.get(url, params=params)
print(response.status_code)
print(json.dumps(response.json(), indent=2))
```
