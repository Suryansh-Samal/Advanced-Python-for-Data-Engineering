```python
from pydantic import BaseModel, Field, EmailStr
from typing import Literal, Optional

class employee_schema(BaseModel):
    name: str = Field(..., min_length=1, max_length=50)
    age: int = Field(..., gt=18, lt=65)
    email: EmailStr = Field(..., description="Employee's email address")
    department: Optional[str] = None
    gender: Literal['Male', 'Female', 'Other']
```

```python
employee = employee_schema(name="John Doe", age=30, email="john.doe@example.com", department="Engineering", gender='Male')
employee
```

# Serialization

```python
employee.model_dump()['name']  # Convert to dictionary
```

```python
employee.model_dump().keys()  # Convert to dictionary
```

```python
employee.model_dump_json()  # Convert to JSON
```
