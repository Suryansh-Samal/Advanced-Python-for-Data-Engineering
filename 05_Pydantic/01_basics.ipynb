```python
from pydantic import BaseModel
```

```python
class employee_schema(BaseModel):
    name: str
    age: int
    department: str
```

```python
employee = employee_schema(name="John Doe", age=30, department="Engineering")
employee
```

## **Coercion VS Strict Types**

# Coercion

### Means pydantic will try its best to run the code by changing the data type if it can like converting the string to int

```python
employee2 = employee_schema(name="Jane Smith", age='25', department="Marketing")
employee2
```

# Strict Types

```python
from pydantic import StrictInt
```

```python
class employee_schema_strict(BaseModel):
    name: str
    age: StrictInt
    department: str
```

```python
employee = employee_schema_strict(name="John Doe", age=30, department="Engineering")
employee
```
